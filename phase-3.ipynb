### Advance Operations With Business Examples

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato Sales Analysis ====")
print("Sales Data Shape: ", sales_data.shape)
print("First 3 restaurant Sales Data\n", sales_data[:3])
print("Restaurant Sales Data\n", sales_data[::3 ])
print("Restaurant Sales Data\n", sales_data[:, 1:]) # : represents all rows and :1 represents column but minus first column
```

```python
# Total Sales Per Year
print(np.sum(sales_data, axis=0)) # axis=0 means columns wise computation 
yearly_total = np.sum(sales_data[:,1:], axis=0)
total = np.sum(yearly_total, axis=0)
print("Yearly Total: ", yearly_total)
print("Total: ", total)

# Minimum Sales Per Restaurant
min_sales = np.min(sales_data[:, 1:], axis=1)
print("Minimum Sales: ", min_sales)

# Maximum Sales Per Year
max_sales_per_year = np.max(sales_data[:, 1:], axis=0)
print("max_sales_per_year: ", max_sales_per_year)

# Average Sales Per Year
avg_sales_per_res = np.mean(sales_data[:, 1:], axis=1) # axis=1 means rows wise computation
print("avg_sales_per_res: ", avg_sales_per_res)

# Cumlative Sales
cumlative_sum = np.cumsum(sales_data[:, 1:], axis=1)
print("-----------------------------------")
print("Cumlative Sales: \n", cumlative_sum)
```

```python
plt.Figure(figsize=(8,6))
plt.plot(np.mean(cumlative_sum, axis=0))
plt.title("Avg Cumlative Sales Across All Restaurant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vector Addition: ", vector1 + vector2)
print("Vector Multiplication: ", vector1 * vector2)
print("Dot Product: ", np.dot(vector1, vector2))
angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle: ", angle)
```
